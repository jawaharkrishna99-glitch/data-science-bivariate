```python
import pandas as pd
data=pd.read_csv("PrePlacement.csv")
data
```

```python
### Statement : Test whether the SSC_p in male and Hsc_p im male are same?
```

```python
## Null Hypothesis: There is no similarity  between 10th pass mark of male 12 th pass mark of male
```

```python
# T Test
from scipy.stats import ttest_rel
male=data[data['gender']=='M']['ssc_p']
male1=data[data['gender']=='M']['hsc_p']
ttest_rel(male,male1)
```

```python
# Reject the hypothesis when p value < 0.05%
# Accept Null hypothesis and Reject the alternative hypothesis
```

```python
### ANOVA

import scipy.stats as stats
stats.f_oneway(data['ssc_p'],data['hsc_p'],data['degree_p'])
```
