```python
import pandas as pd
import numpy as np
```

```python
import io
# Dynamically find the uploaded filename containing 'PrePlacement'
filename = next(k for k in uploaded.keys() if 'PrePlacement' in k and k.endswith('.csv'))
dataset = pd.read_csv(io.BytesIO(uploaded[filename]))
dataset
```

```python

from statsmodels.stats.outliers_influence import variance_inflation_factor

def calc_vif(x):
    vif=pd.DataFrame()
    vif['variables']=x.columns
    vif["VIF"]=[variance_inflation_factor(x.values,i) for i in range(x.shape[1])]

    return(vif)
```

```python

calc_vif(dataset[["ssc_p","hsc_p","degree_p","etest_p","salary"]])
```
