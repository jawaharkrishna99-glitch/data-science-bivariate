```python
import os
import pandas as pd

# Search for the CSV file in the current workspace
target_file = 'Pre placement.csv'
file_path = None

for root, dirs, files in os.walk('.'):
    if target_file in files:
        file_path = os.path.join(root, target_file)
        break

if file_path:
    print(f"Found file at: {file_path}")
    data = pd.read_csv(file_path)
    display(data.head())
else:
    print(f"Could not find '{target_file}' in the current workspace directory tree. Please upload it or verify its name.")
```

```python
from google.colab import files
import io
import pandas as pd

print("Please select and upload 'Pre placement.csv' from your computer:")
uploaded = files.upload()

if 'Pre placement.csv' in uploaded:
    data = pd.read_csv(io.BytesIO(uploaded['Pre placement.csv']))
    print("File successfully loaded into 'data' variable!")
    display(data.head())
else:
    # Fallback in case of a different capitalization or name
    uploaded_filename = list(uploaded.keys())[0]
    data = pd.read_csv(io.BytesIO(uploaded[uploaded_filename]))
    print(f"Loaded {uploaded_filename} into 'data' variable!")
    display(data.head())
```

```python

data=pd.get_dummies(data).astype(int)
data
```
